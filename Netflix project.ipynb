```python
df.drop_duplicates(inplace=True)
```

```python
df.duplicated().sum()
```

```python
# Dirty Values - N/A, NULL, Blank, Spaces

#df.replace(
#    ['N/A','NULL','',' '],np.nan,inplace=True
#)

for col in df.select_dtypes(include='object').columns:
    df[col] = df[col].str.strip()
```

```python
df.replace(
    ['N/A', 'NULL', 'NA', 'null', 'n/a', ''],
    np.nan,
    inplace=True
)
```

```python
# Extra Spaces Remove

df = df.apply(lambda x:x.str.strip() if x.dtype == "object" else x )
```

```python
# Standardize Text

df["Country"] = df["Country"].str.title()

df["Genre"] = df["Genre"].str.title()

df["Subscription_Plan"] = (df["Subscription_Plan"].str.title())

df["Account_Status"] = (df["Account_Status"].str.title())
```

```python
df["Country"].value_counts()
```

```python
# Numeric Columns

numeric_cols = ['Age','Watch_Hours','Monthly_Fee','Movies_Watched','Episodes_Watched','Avg_Rating','Monthly_Sessions','Download_Count']
```

```python
for col in numeric_cols:
    df[col] = pd.to_numeric(df[col], errors='coerce')
```

```python
df.describe()
```

```python
df =df[(df['Age']>=18)&(df['Age']<=100)]
```

```python
df['Age'].describe()
```

```python
# Negative Values

df = df[df['Watch_Hours']>=0]
df = df[df['Monthly_Fee']>=0]
```

```python
# Data Cleaning

df['Join_Date'] = pd.to_datetime(df['Join_Date'],errors = 'coerce')
df['Last_Watch_Date'] = pd.to_datetime(df['Last_Watch_Date'],errors = 'coerce')
```

```python
df[["Join_Date","Last_Watch_Date"]].isnull().sum()
```

```python
# Handle Missing Values

df["Age"]=df["Age"].fillna(
    df["Age"].mean()
)

df["Watch_Hours"]=df["Watch_Hours"].fillna(
    df["Watch_Hours"].mean()   
)

df["Monthly_Fee"]=df["Monthly_Fee"].fillna(
    df["Monthly_Fee"].mean(),
)
```

```python
df["Payment_Method"]=df["Payment_Method"].fillna("Unknown")

df["Device"]=df["Device"].fillna("Unknown")

df["Genre"]=df["Genre"].fillna("Unknown")
```

```python
# Feature Engineering

# Customer Revenue

df["Customer_Revenue"] = (df["Monthly_Fee"] *df["Monthly_Sessions"])
```

```python
df.head()
```

```python
# Engagement Score

df["Engagement_Score"] = (df["Watch_Hours"] +df["Monthly_Sessions"] +df["Download_Count"])
```

```python
df.head()
```

```python
# Watch Category

bins = [0,20,50,100,np.inf]
label = ['Low','Medium','High','Very Heigh']

df['Watch_Category'] =pd.cut(df['Watch_Hours'],bins = bins , labels = label)
```

```python
df.head()
```

```python
df['Watch_Category'].value_counts()
```

```python
# Customer Age Group

bins = [17,25,35,50,100]
label = ['18-25','26-35','36-50','51+']

df['Age_Group'] = pd.cut(df['Age'],bins = bins, labels = label)
```

```python
df.head()
```

```python
df['Age_Group'].value_counts()
```

```python
# Account Active Flag
df["Active_Flag"] = np.where(df["Account_Status"] == "Active",1,0)
```

```python
df.head()
```

```python
# Final Data Quality Check
df.info()
```

```python
# Clean csv Export
df.to_csv('Netflix_Clean_Data.csv',index = False)
```

```python
import os
import glob
import pandas as pd
from sqlalchemy import create_engine
import urllib

# 1. Current working directory मधून CSV शोधा (no hardcoded path)
csv_files = glob.glob('*.csv')
print("Current folder:", os.getcwd())
print("CSV files found:", csv_files)

if not csv_files:
    raise FileNotFoundError("No CSV files found in current directory")

csv_path = csv_files[0]
print("Using file:", csv_path)

# 2. Read CSV
df = pd.read_csv(csv_path)
print("Rows read from CSV:", len(df))

# 3. Connect to SQL Server (Windows Authentication)
server = r'localhost\SQLEXPRESS'
database = 'netflix_db'

params = urllib.parse.quote_plus(
    f"DRIVER={{ODBC Driver 17 for SQL Server}};"
    f"SERVER={server};"
    f"DATABASE={database};"
    f"Trusted_Connection=yes;"
)
engine = create_engine(f"mssql+pyodbc:///?odbc_connect={params}", fast_executemany=True)

# 4. Write to SQL
df.to_sql(
    'netflix_customers',
    con=engine,
    if_exists='replace',
    index=False
)

# 5. Verify
with engine.connect() as conn:
    count = conn.exec_driver_sql("SELECT COUNT(*) FROM netflix_customers").fetchone()[0]
    print("Rows in SQL table:", count)
```
